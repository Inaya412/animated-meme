5_Values in Python should usually be named in "lower snake case". What does this mean? Include some examples._

Lower snake case means using lowercase letters and separating words with underscores (`_`).

Examples include `first_name`, `total_runs`, `exam_score` and `price_per_item`.

It makes variable names easier to read and understand.



_But some values will be named in "SCREAMING SNAKE CASE". Really. That's what it's called. What does this tell us about the value?_

SCREAMING SNAKE CASE means using capital letters and separating words with underscores (`_`). It usually tells us that the value is a constant, meaning a value that should not change while the program runs.

Examples include `MAX_SCORE`, `PI` and `TAX_RATE`.

_Jot down three, or four, factors that should be considered when choosing a name for a value._

1. **Meaningful:** The name should describe what the value represents, for example `total_score`.
2. **Readable:** Use clear names that are easy to understand.
3. **Naming style:** Use lower snake case for variables, such as `first_name`.
4. **Consistency:** Use the same naming style throughout the program.

_Should you add a space before the bracket after a `print` command? Why?_

No, you should not add a space between `print` and the opening bracket. In Python, the correct style is `print("Hello")`, because `print` is a function and the brackets should follow its name directly.


_What about spaces around an expression? `2 + 2` or `2+2`?_

You should use spaces around operators, so `2 + 2` is preferred over `2+2`. This makes the code easier to read and is the recommended Python style.

_What purpose does a blank line serve in a Python program?_

A blank line separates different parts of a program, making the code easier to read and understand. It does not affect how the program runs.


_What should be on the last line of every Python program?_

The last line of a Python program should not be blank. The file should end with a newline character after the final line of code.


_What is a program comment? Why are these almost always best avoided?_

A program comment is a note in the code that Python ignores. Comments start with a `#` symbol. They are often best avoided because good variable names and clear code should explain what the program does. However, comments are useful when explaining something that is not obvious.

_Python programmers often refer to `PEP-8`, a much revered document. What is it? (Google is your friend here.)_

PEP 8 stands for Python Enhancement Proposal 8. It is the official style guide for writing Python code. It gives programmers guidelines for naming variables, using spaces and indentation, and organising code so that it is easy to read and understand.

_Suppose you write a program, and the person sat next to you writes the same program. Assuming both work, would they be exactly the same? Why (not)?_

No, the programs would not necessarily be exactly the same. There are often many different ways to write a program that produces the same result. We might use different variable names, prompts or calculations, but both programs could still work correctly.

_AI-generated program code is very easy to spot. Can you work out any features it might have that makes this so?

AI-generated code may be easy to spot because it can have overly detailed comments, long explanations, generic variable names and unnecessary code. It may also use a different coding style from the rest of the program. However, these features do not always mean that code was written by AI._
